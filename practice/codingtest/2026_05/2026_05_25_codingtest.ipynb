# 👩‍💻 01. 최대공약수와 최소공배수

두 수를 입력받아 두 수의 최대공약수와 최소공배수를 반환하는 함수, solution을 완성해 보세요. 배열의 맨 앞에 최대공약수, 그다음 최소공배수를 넣어 반환하면 됩니다. 예를 들어 두 수 3, 12의 최대공약수는 3, 최소공배수는 12이므로 solution(3, 12)는 [3, 12]를 반환해야 합니다.

```python
n = 3
m = 12
```

```python
answer = []
```

```python
# n의 최소공배수
n_LCM = []
for i in range(1, n+1, 1):
    result = n * i
    n_LCM.append(result)
print(f"{n}의 공배수는 {n_LCM}...")

# m의 최소공배수
m_LCM = []
for i in range(1, m+1, 1):
    result = m * i
    m_LCM.append(result)
print(f"{m}의 공배수는 {m_LCM}...")
```

```python
### 최대공약수 구하기 ###
# 유클리드 호제법: 두 양의 정수 혹은 두 다항식의 최대공약수를 구하는 알고리즘이다.
# 180과 108의 최대공약수
# 180 ＝ 108×1 ＋ 72
# 108 ＝  72×1 ＋ 36
#  72 ＝  36×2 ＋  0
# 최대공약수는 36이다.
```
