# 💻 Q1. 부분 문자열
어떤 문자열 A가 다른 문자열 B안에 속하면 A를 B의 부분 문자열이라고 합니다. 예를 들어 문자열 "abc"는 문자열 "aabcc"의 부분 문자열입니다.

문자열 str1과 str2가 주어질 때, str1이 str2의 부분 문자열이라면 1을 부분 문자열이 아니라면 0을 return하도록 solution 함수를 완성해주세요.

```python
str1 = "abc"
str2 = "aabcc"
```

```python
str1 = "tbt"
str2 = "tbbttb"
```

```python
if str1 in str2:
    print("1")
else:
    print("0")
```

# 💻 Q2. 조건 문자열
문자열에 따라 다음과 같이 두 수의 크기를 비교하려고 합니다.


- 두 수가 n과 m이라면
- ">", "=" : n >= m
- "<", "=" : n <= m
- ">", "!" : n > m
- "<", "!" : n < m

두 문자열 ineq와 eq가 주어집니다. ineq는 "<"와 ">"중 하나고, eq는 "="와 "!"중 하나입니다. 그리고 두 정수 n과 m이 주어질 때, n과 m이 ineq와 eq의 조건에 맞으면 1을 아니면 0을 return하도록 solution 함수를 완성해주세요.

```python
ineq = "<"
eq = "="
n = 20
m = 50
```

```python
ineq = ">"
eq = "!"
n = 41
m = 78
```

```python
# 경우의 수
# <=, >=, >, < 4가지
#if <
#     만약 =
#         만약 n <= m:
#             1
#         아니면:
#             0

#     만약 ! 이고
#         만약 n < m 이면:
#             1
#         혹은
#             0

# else >
#     만약 =
#         만약 n >= m:
#             1
#         아니면:
#             0

#     만약 ! 이고
#         만약 n > m 이면:
#             1
#         혹은
#             0
```

```python
if ineq == "<":
    if eq == "=":
        if (n <= m):
            print(1)
        else:
            print(0)
    elif eq == "!":
        if(n < m):
            print(1)
        else:
            print(0)


else:
    if eq == "=":
        if(n >= m):
            print(1)
        else:
            print(0)
    elif eq == "!":
        if (n > m):
            print(1)
        else:
            print(0)

```
