# 👩‍💻 02. 내적

길이가 같은 두 1차원 정수 배열 a, b가 매개변수로 주어집니다. a와 b의 내적을 return 하도록 solution 함수를 완성해주세요.

이때, a와 b의 내적은 a[0]*b[0] + a[1]*b[1] + ... + a[n-1]*b[n-1] 입니다. (n은 a, b의 길이)



```python
a = [1,2,3,4]
b = [-3,-1,0,2]
```

```python
a = [-1,0,1]
b = [1,0,-1]
```

```python
answer = []
for a, b, in zip(a,b):
    c = a * b
    answer.append(c)

print(answer)

print(sum(answer))
```

```python
def solution(a,b):
    answer = []
    for a, b, in zip(a,b):
        c = a * b
        answer.append(c)



    return sum(answer)
```

```python
solution(a,b)
```

단어의 길이가 짝수라면 가운데 두글자를 반환하면 됩니다.


```python
s = "abcde"
```

```python
# s의 길이가 홀수라면
    # 가운데 글자 출력(ceil)
# 짝수라면
    # 가운데 두 글자 출력

num_len = len(s)
print(num_len)

import math
num = len(s)/2
print(num)

a = math.ceil(num)
print(a)

print(s[a-1:a+1])
print(s[a-1])
```

```python
import math
num_len = len(s)
a = math.ceil(num_len/2)

if num_len % 2 == 0: # 짝수
    print(s[a-1:a+1])
else:  # 홀수
    print(s[a-1])
```

```python
def solution(s):
    import math
    num_len = len(s)
    a = math.ceil(num_len/2)

    if num_len % 2 == 0: # 짝수
        return s[a-1:a+1]
    else:  # 홀수
        return s[a-1]
```

# 👩‍💻 04. 같은 숫자는 싫어

배열 arr가 주어집니다. 배열 arr의 각 원소는 숫자 0부터 9까지로 이루어져 있습니다.
이때, 배열 arr에서 연속적으로 나타나는 숫자는 하나만 남기고 전부 제거하려고 합니다. 단, 제거된 후 남은 수들을 반환할 때는 배열 arr의 원소들의 순서를 유지해야 합니다.

 예를 들면,

arr = [1, 1, 3, 3, 0, 1, 1] 이면 [1, 3, 0, 1] 을 return 합니다.


arr = [4, 4, 4, 3, 3] 이면 [4, 3] 을 return 합니다.


배열 arr에서 연속적으로 나타나는 숫자는 제거하고 남은 수들을 return 하는 solution 함수를 완성해 주세요.

```python
arr = [1,1,3,3,0,1,1]
```

```python
answer = []
for i in arr:
    if i not in answer:
        answer.append(i)
print(answer)
```
