# 👩‍💻 01. 문자열 내 p와 y의 개수

대문자와 소문자가 섞여있는 문자열 s가 주어집니다. s에 'p'의 개수와 'y'의 개수를 비교해 같으면 True, 다르면 False를 return 하는 solution를 완성하세요. 'p', 'y' 모두 하나도 없는 경우는 항상 True를 리턴합니다. 단, 개수를 비교할 때 대문자와 소문자는 구별하지 않습니다.

예를 들어 s가 "pPoooyY"면 true를 return하고 "Pyy"라면 false를 return합니다.

> 제한사항
- 문자열 s의 길이 : 50 이하의 자연수
- 문자열 s는 알파벳으로만 이루어져 있습니다.


```python
s = "pPoooyY"
```

```python
s = "Pyy"
```

```python
# 대소문자 구별 안함
lower = s.lower()
# p의 개수 y의 개수 비교

print(lower.count('p'))
print(lower.count('y'))

# 같으면 True 다르면 False
```

```python
count_p = lower.count('p')
count_y = lower.count('y')

if count_p == count_y:
    print("True")
else:
    print("False")
```

```python
def solution(s):
    lower = s.lower()
    count_p = lower.count('p')
    count_y = lower.count('y')

    if count_p == count_y:
        return True
    else:
        return False
```

```python
s = "pPoooyY"
solution(s)
```

```python
s = "Pyy"
solution(s)
```

# 👩‍💻 02. 음양 더하기

어떤 정수들이 있습니다. 이 정수들의 절댓값을 차례대로 담은 정수 배열 absolutes와 이 정수들의 부호를 차례대로 담은 불리언 배열 signs가 매개변수로 주어집니다. 실제 정수들의 합을 구하여 return 하도록 solution 함수를 완성해주세요.

```python
absolutes = [4,7,12]
signs = [True,False,True]
```

```python
# absolutes[0]
# signs[0]

# signs가 True 라면
#     absolutes 는 그대로
# 아니라면 False:
#     absolutes는 -1곱해

# answer에 입력해놔

# answer.sum 결과 출력
```

```python
for i in range(len(absolutes)):
    if signs[i] == True:

        print(absolutes[i])
    else:
        print(-1 * absolutes[i])
```

```python
result = []
for i in range(len(absolutes)):
    if signs[i] == True:

        positive = (absolutes[i])
        result.append(positive)
    else:
        negative = (-1 * absolutes[i])
        result.append(negative)

print(result)

answer = sum(result)
print(answer)
```

```python
def solution(absolutes, signs):
    result = []
    for i in range(len(absolutes)):
        if signs[i] == True:

            positive = (absolutes[i])
            result.append(positive)
        else:
            negative = (-1 * absolutes[i])
            result.append(negative)

    answer = sum(result)
    return answer
```

```python
absolutes = [1,2,3]
signs = [False, False, True]
solution(absolutes, signs)
```

# 👩‍💻 03. 없는 숫자 더하기
0부터 9까지의 숫자 중 일부가 들어있는 정수 배열 numbers가 매개변수로 주어집니다.

Numbers에서 찾을 수 없는 0부터 9까지의 숫자를 모두 찾아 더한 수를 return 하도록 solution 함수를 완성해주세요.

```python
numbers = [1,2,3,4,6,7,8,0]
```

```python
# 0 ~ 9 까지 넣어
# numbers에 없다면 print
# sum 하고 return

answer = []
for i in range(0,10,1):
    if i not in numbers:
        answer.append(i)
        print(answer)
print(sum(answer))
```

```python
def solution(numbers):
    answer = []
    for i in range(0,10,1):
        if i not in numbers:
            answer.append(i)

    return (sum(answer))
```

```python
numbers = [5,8,4,0,6,7,9]
solution(numbers)
```

# 👩‍💻 04. 나누어 떨어지는 숫자 배열
array의 각 element 중 divisor로 나누어 떨어지는 값을 오름차순으로 정렬한 배열을 반환하는 함수, solution을 작성해주세요.

divisor로 나누어 떨어지는 element가 하나도 없다면 배열에 -1을 담아 반환하세요.

```python
arr = [5, 9, 7, 10]
divisor = 5
```

```python
arr = [2, 36, 1, 3]
divisor = 1
```

```python
arr = [3, 2, 6]
divisor = 10
```

```python
# arr에 있는 거 다 입력해봐

# arr 을 divisor로 나눠
# 만약 나머지가 0 이면
# answer에 붙여

# 오름차순 정렬
# 만약 아무것도 안 붙었어 -> -1 출력
```

```python
answer = []
for i in range(len(arr)):
    if arr[i] % divisor == 0:
        answer.append(arr[i])
        answer.sort()

if len(answer) == 0:
    answer = [-1]

print(answer)
```

```python
def solution(arr,divisor):
    answer = []
    for i in range(len(arr)):
        if arr[i] % divisor == 0:
            answer.append(arr[i])
            answer.sort()

    if len(answer) == 0:
        answer = [-1]

    return answer
```

```python
solution(arr,divisor)
```
