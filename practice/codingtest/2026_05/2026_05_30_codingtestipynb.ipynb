# 👩‍💻 01. x만큼 간격이 있는 n개의 숫자

함수 solution은 정수 x와 자연수 n을 입력 받아, x부터 시작해 x씩 증가하는 숫자를 n개 지니는 리스트를 리턴해야 합니다. 다음 제한 조건을 보고, 조건을 만족하는 함수, solution을 완성해주세요.

```python
x = 2
n = 5
```

```python
x = 4
n = 3
```

```python
x = -4
n = 2
```

```python
x * n
```

```python
answer = []
for i in range(1, n+1):
    answer.append(x * i)
print(answer)
```

```python
answer = [(x * i) for i in range(1, n+1)]
print(answer)
```

```python
def solution(x, n):
    answer = [(x * i) for i in range(1, n+1)]
    return answer
```
