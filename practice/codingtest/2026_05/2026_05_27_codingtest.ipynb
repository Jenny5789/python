```python
def solution(s):
    answer = []
    word = s.split(" ")
    for i in range(len(word)):

        first = word[i][0]

        if first.isdigit():
            first_word = first + word[i][1:]
            fwc = first_word.capitalize()
            answer.append(fwc)

        else:
            first = first.title()
            first_word = first + word[i][1:]
            fwc = first_word.capitalize()
            answer.append(fwc)

        rea_answer = ' '.join(answer)
    return rea_answer
```
