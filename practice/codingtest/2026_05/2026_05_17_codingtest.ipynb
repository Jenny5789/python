# 👩‍💻 01. 나머지가 1이 되는 수 찾기

자연수 n이 매개변수로 주어집니다. n을 x로 나눈 나머지가 1이 되도록 하는 가장 작은 자연수 x를 return 하도록 solution 함수를 완성해주세요. 답이 항상 존재함은 증명될 수 있습니다.



```python
n = 10
```

```python
def solution(n):
    answer = 0
    for i in range(1,n,1):
        if n % i == 1:
            answer = i
            break

    return answer
```

```python
solution(10)
```

# 👩‍💻 02. 두 정수 사이의 합
두 정수 a, b가 주어졌을 때 a와 b 사이에 속한 모든 정수의 합을 리턴하는 함수, solution을 완성하세요.
예를 들어 a = 3, b = 5인 경우, 3 + 4 + 5 = 12이므로 12를 리턴합니다.


```python
a = 3
b = 5
```

```python
def solution(a,b):
    number = []
    answer = 0

    if a < b:

        for i in range(a,b+1,1):
            number.append(i)
        answer = sum(number)

    elif a > b:
        for i in range(b,a+1,1):
            number.append(i)
        answer = sum(number)

    else:
        answer = a

    return answer
```

```python
print(solution(3,5))
print(solution(5,3))
print(solution(5,5))
```
