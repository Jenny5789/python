```python
answer = ''
watermelon = "수박"

if n % 2 == 1:
    rep = n / 2 + 1
else:
    rep = n /2

word_list = int(rep) * watermelon

print( word_list[:n])
```

```python
def solution(n):

    answer = ''
    watermelon = "수박"

    if n % 2 == 1:
        rep = n / 2 + 1
    else:
        rep = n /2

    word_list = int(rep) * watermelon

    return word_list[:n]
```

```python
solution(n)
```

# 👩‍💻 02. 정수 제곱근 판별

임의의 양의 정수 n에 대해, n이 어떤 양의 정수 x의 제곱인지 아닌지 판단하려 합니다.

n이 양의 정수 x의 제곱이라면 x+1의 제곱을 리턴하고, n이 양의 정수 x의 제곱이 아니라면 -1을 리턴하는 함수를 완성하세요.

```python
n = 121
```

```python
n = 3
```

```python
# 양의 제곱근 확인
# 그렇다면 (n + 1) ^ 2
# 양의 정수가 아니라면 -1 반납
```

```python
import math
a = math.sqrt(n)
print(a)
if  a % 1 == 0 :
    if a >= 0:
        a += 1
        print(int(a * a))
    else:
        print(-1)
else:
    print(-1)
```

```python
def solution(n):
    import math
    a = math.sqrt(n)

    if  a % 1 == 0 :
        if a >= 0:
            a += 1
            return (int(a * a))
        else:
            return -1
    else:
        return -1
```

```python
solution(n)
```
