# 👩‍💻 Q2. 짝수와 홀수

정수 num이 짝수일 경우 "Even"을 반환하고 홀수인 경우 "Odd"를 반환하는 함수, solution을 완성해주세요.

제한 조건
- num은 int 범위의 정수입니다.
- 0은 짝수입니다.

```python
num = 3
```

```python
num = 4
```

```python
answer = ''
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
def solution(num):
    answer = ''
    if num % 2 == 0:
        answer = "Even"
    else:
        answer = "Odd"


    return answer
```

```python
print(solution(3))
print(solution(4))
```

# 👩‍💻 Q4. 하샤드 수
양의 정수 x가 하샤드 수이려면 x의 자릿수의 합으로 x가 나누어져야 합니다.
예를 들어 18의 자릿수 합은 1+8=9이고, 18은 9로 나누어 떨어지므로 18은 하샤드 수입니다. 자연수 x를 입력받아 x가 하샤드 수인지 아닌지 검사하는 함수, solution을 완성해주세요.

```python
x = 10
```

```python
x = 11
```

```python
# x의 자리수 합(sum)구하기
# x이 sum으로 나누어 떨어지는지 검사
# 만약 떨어진다면 true
# 아니라면 false

x_sum = sum(map(int, str(x)))
print(x_sum)

if x % x_sum == 0:
    print("True")
else:
    print("False")
```

```python
def solution(x):
    x_sum = sum(map(int, str(x)))
    answer = x_sum

    if x % x_sum == 0:
        answer = True
    else:
        answer = False

    return answer
```

```python
print(solution(10))
print(solution(11))
```

```python
# x*n 이 양수일 때/ 음수일 때 구분
```

```python
abs(x * n) + 1
```

```python
def solution(x, n):
    answer = []
    if x * n >= 0:
        answer = list(range(x, (x * n) + 1, x))

    else:
        answer = list(range(x, (x * n) - 1, x))

    return answer
```

```python
print(solution(2,5))
print(solution(-4,2))
```

# 👩‍💻 Q1. 정수 제곱근 판별
임의의 양의 정수 n에 대해, n이 어떤 양의 정수 x의 제곱인지 아닌지 판단하려 합니다.

n이 양의 정수 x의 제곱이라면 x+1의 제곱을 리턴하고, n이 양의 정수 x의 제곱이 아니라면 -1을 리턴하는 함수를 완성하세요.

```python
n = 121
```

```python
n = 3
```

```python
def solution(n):
    import math
    answer = 0
    # x 구하기
    x = math.sqrt(n)   # float로 반환

    # n 판단한기
    # n이 양수 정수 x의 제곱이라면
    if x is int:
        # x + 1 리턴
        answer = (x+1)
    # 아니라면
    else:
        # -1 리턴
        answer = -1

    return answer
```
