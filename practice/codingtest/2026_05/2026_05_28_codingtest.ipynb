# 👩‍💻 01. JadenCase 문자열 만들기

JadenCase란 모든 단어의 첫 문자가 대문자이고, 그 외의 알파벳은 소문자인 문자열입니다.

 단, 첫 문자가 알파벳이 아닐 때에는 이어지는 알파벳은 소문자로 쓰면 됩니다. (첫 번째 입출력 예 참고)

문자열 s가 주어졌을 때, s를 JadenCase로 바꾼 문자열을 리턴하는 함수, solution을 완성해주세요.

```python
s = "3people unFollowed me"
```

```python
s = "for the last week"
```

```python
s = "3people unFollowed     me"
```

```python
s = "people unFoLlowed     1me"
```

```python
s = "people 3unFoLlowed     1me"
```

```python
s = "  3people   unFollowed  "
```

```python
def solution(s):
    answer = []
    space = ' '.join(s.split())

    text = space.strip()

    word = text.split(' ')
    for i in range(len(word)):

        first = word[i][0]

        if first.isdigit():
            first_word = first + word[i][1:]
            fwc = first_word.capitalize()
            answer.append(fwc)

        else:
            first = first.title()
            first_word = first + word[i][1:]
            fwc = first_word.capitalize()
            answer.append(fwc)

        rea_answer = ' '.join(answer)
    return rea_answer
```

# 👩‍💻 02. 올바른 괄호

괄호가 바르게 짝지어졌다는 것은 '(' 문자로 열렸으면 반드시 짝지어서 ')' 문자로 닫혀야 한다는 뜻입니다. 예를 들어

- "()()" 또는 "(())()" 는 올바른 괄호입니다.
- ")()(" 또는 "(()(" 는 올바르지 않은 괄호입니다.

'(' 또는 ')' 로만 이루어진 문자열 s가 주어졌을 때, 문자열 s가 올바른 괄호이면 true를 return 하고, 올바르지 않은 괄호이면 false를 return 하는 solution 함수를 완성해 주세요.

```python
s = "()()"
```

```python
s = "(())()"
```

```python
s = ")()("
```

```python
s = "(()("
```

```python
answer = []
stack = s
is_valid = True

for i in stack:
    if i == "(":
        answer.append(i)

    elif i == ")":

        if len(answer) == 0:
            is_valid =  False
            break

        answer.pop()

    else:
        break


if len(answer) != 0:
    is_valid = False

print(is_valid)
```

```python
def solution(s):
    answer = []
    stack = s
    is_valid = True

    for i in stack:
        if i == "(":
            answer.append(i)

        elif i == ")":

            if len(answer) == 0:
                is_valid =  False
                break

            answer.pop()

        else:
            break


    if len(answer) != 0:
        is_valid = False

    return is_valid
```
