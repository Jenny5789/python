# 💻 Q1. 문자열 섞기
길이가 같은 두 문자열 str1과 str2가 주어집니다.

두 문자열의 각 문자가 앞에서부터 서로 번갈아가면서 한 번씩 등장하는 문자열을 만들어 return 하는 solution 함수를 완성해 주세요.

```python
str1 = "aaaaa"
str2 = "bbbbb"
```

```python
answer = ''
for char1, char2 in zip(str1, str2):
    answer += char1 + char2
print(answer)
```

# 💻 Q2. 카운트 다운
정수 start_num와 end_num가 주어질 때,

start_num에서 end_num까지 1씩 감소하는 수들을 차례로 담은 리스트를 return하도록 solution 함수를 완성해주세요.

```python
start_num = 10
end_num = 3
```

```python
answer = list(range(10,2,-1))
print(answer)
```

```python
answer = list(range(start_num,end_num-1,-1))
print(answer)
```

# 💻 Q3. 배열 만들기 1
정수 n과 k가 주어졌을 때,

 1 이상 n이하의 정수 중에서 k의 배수를 오름차순으로 저장한 배열을 return 하는 solution 함수를 완성해 주세요.

```python
n = 10
k = 3
```

```python
n = 15
k = 5
```

```python
answer = []
for i in range(1, n + 1):
    if i % k == 0:
        answer.append(i)
print(answer)
```

# 💻 Q4. 9로 나눈 나머지
음이 아닌 정수를 9로 나눈 나머지는 그 정수의 각 자리 숫자의 합을 9로 나눈 나머지와 같은 것이 알려져 있습니다.


이 사실을 이용하여 음이 아닌 정수가 문자열 number로 주어질 때, 이 정수를 9로 나눈 나머지를 return 하는 solution 함수를 작성해주세요.

```python
number = "123"
```

```python
number = "78720646226947352489"
```

```python
answer = 0
for i in number:
    answer += int(i)
    answer = answer % 9
print(answer)
```
