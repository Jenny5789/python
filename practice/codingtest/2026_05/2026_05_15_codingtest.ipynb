# 👩‍💻 Q2. 약수의 합

정수 n을 입력받아 n의 약수를 모두 더한 값을 리턴하는 함수, solution을 완성해주세요.

제한 사항
- n은 0 이상 3000이하인 정수입니다.

```python
n = 12
```

```python
n = 5
```

```python
answer = 0
com_fac = []
# 약수 i 부터 n-1까지 넣어 n이 나눈 나머지가 0이면 약수
for i in range(1,(n)):
    if n % i == 0:
        com_fac.append(i)

com_fac.append(n)
print(com_fac)
```

```python
answer = sum(com_fac)
print(answer)
```

# 👩‍💻 Q3. 자릿수 더하기

# 👩‍💻 Q5. 정수 내림차순으로 배치하기
함수 solution은 정수 n을 매개변수로 입력받습니다. n의 각 자릿수를 큰것부터 작은 순으로 정렬한 새로운 정수를 리턴해주세요. 예를들어 n이 118372면 873211을 리턴하면 됩니다.


```python
n = 118372
```

```python
answer = []
for i in str(n):
    answer.append(i)
print(answer)
```

```python
dec = sorted(answer, reverse = True)
print(dec)
```

```python
result_str = ""
for i in dec:
    result_str += i

print(result_str)
```
