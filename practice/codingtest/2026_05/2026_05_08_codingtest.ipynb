# 5/6 문제->  💻 Q3. ad 제거하기
문자열 배열 strArr가 주어집니다.

배열 내의 문자열 중 "ad"라는 부분 문자열을 포함하고 있는 모든 문자열을 제거하고

남은 문자열을 순서를 유지하여 배열로 return 하는 solution 함수를 완성해 주세요.

```python
strArr = ["and","notad","abcd"]
```

```python
strArr = ["there","are","no","a","ds"]
```

```python
answer = []
for i in strArr:
    if "ad" in i:
        strArr.remove(i)
    else:
        answer.append(i)

print(answer)
```

# 💻 Q4. 홀짝에 따라 다른 값 반환하기

양의 정수 n이 매개변수로 주어질 때,

 n이 홀수라면 n 이하의 홀수인 모든 양의 정수의 합을 return 하고

 n이 짝수라면 n 이하의 짝수인 모든 양의 정수의 제곱의 합을 return 하는 solution 함수를 작성해 주세요.

```python
n = 7
```

```python
n = 10
```

```python
answer = 0
if n % 2 == 1:
    for i in range(1,n+1,2):
        answer += i
else:
    for i in range(0, n+1, 2):
        sq = i * i
        answer += sq
print(answer)
```

# 💻 Q5. 배열에서 문자열 대소문자 변환하

자열 배열 strArr가 주어집니다.

모든 원소가 알파벳으로만 이루어져 있을 때, 배열에서 홀수번째 인덱스의 문자열은 모든 문자를 대문자로,

 짝수번째 인덱스의 문자열은 모든 문자를 소문자로 바꿔서 반환하는 solution 함수를 완성해 주세요.

```python
strArr = ["AAA","BBB","CCC","DDD"]
```

```python
strArr = ["aBc","AbC"]
```

```python
answer = []
for i in range(len(strArr)):
    if i % 2 == 1:
        a = strArr[i].upper()
        answer.append(a)
    else:
        b = strArr[i].lower()
        answer.append(b)

print(answer)
```
