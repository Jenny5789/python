myString의 "A"를 "B"로, "B"를 "A"로 바꾼 문자열의 연속하는 부분 문자열 중

 pat이 있으면 1을 아니면 0을 return 하는 solution 함수를 완성하세요.

```python
myString = "ABBAA"
pat = "AABB"
```

```python
myString = "ABAB"
pat = "ABAB"
```

```python
result = myString.replace("A","a").replace("B","A").replace("a","B")
print(result)
```

```python
if pat in result:
    print("1")
else:
    print("0")
```
