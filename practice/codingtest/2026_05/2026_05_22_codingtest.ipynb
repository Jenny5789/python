# 👩‍💻 01. 직사각형 별 찍기
이 문제에는 표준 입력으로 두 개의 정수 n과 m이 주어집니다.
별(*) 문자를 이용해 가로의 길이가 n, 세로의 길이가 m인 직사각형 형태를 출력해보세요.

```python
a = 5
b = 3
```

```python
n = "*" * a
print(n)
```

```python
for i in range(b):
    print(n)
```

```python
a, b = map(int, input().strip().split(' '))
n = "*" * a
for i in range(b):
    print(n)
```

# 👩‍💻 02. 행렬의 덧셈
행렬의 덧셈은 행과 열의 크기가 같은 두 행렬의 같은 행, 같은 열의 값을 서로 더한 결과가 됩니다.

2개의 행렬 arr1과 arr2를 입력받아, 행렬 덧셈의 결과를 반환하는 함수, solution을 완성해주세요.

```python
arr1 = [[1,2],[2,3]]
arr2 = [[3,4],[5,6]]
```

```python
for a, b in zip(arr1, arr2):
    print(a, b)
```

```python
c = [1, 2]
d = [3, 4]
```

```python
for c, d in zip(c,d):
    print(c+d)
```

```python
answer = []
for a, b in zip(arr1, arr2):
    first = []
    print(first)
    for c, d in zip(a, b):
        first.append(c+d)

    answer.append(first)

    print(answer)
```

```python
def solution(arr1, arr2):
    answer = []
    for a, b in zip(arr1, arr2):
        first = []
        print(first)
        for c, d in zip(a, b):
            first.append(c+d)

        answer.append(first)

    return answer
```

```python
solution(arr1, arr2)
```
