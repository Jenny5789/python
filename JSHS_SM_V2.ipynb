* 과목별 점수계산(국 영 수)
* 평균 계산

```python
import random as rand

last_name = ["김", "이", "박", "최", "정", "강", "조", "윤", "장", "임"]
first_name = ["민준", "서준", "도윤", "주원", "예준", "현우", "준서", "지훈", "하준", "우진", "서연", "서윤", "지우", "서현", "민서",
    "하은", "지민", "윤서", "지안", "은서"]


# 초기 DB 생성 함수
def init_db():
    db = {}
    name_index = {}

    for grade in range(1, 4):
        for class_num in range(1, 9):
            last_std_num = rand.randint(25, 30)
            for num in range(1, last_std_num + 1):
                name = rand.choice(last_name) + rand.choice(first_name)
                std_id = (grade, class_num, num)
                db[std_id] = {
                    "이름":name,
                    "학년": grade,
                    "반": class_num,
                    "번호": num,
                    "성적": {"국어": 0, "영어": 0, "수학": 0},
                    "평균": 0.0
                }

                if name not in name_index:
                    name_index[name] = []
                name_index[name].append(std_id)
    return db, name_index


# 학생 추가 함수
def add_student(db, name_index):
    confirm = input("\n학생 정보를 추가하시겠습니까? (Y/N): ").upper()
    if confirm != "Y":
        print("추가가 취소되었습니다.")
        return

    try:
        name = input("이름: ")
        grade = int(input("학년: "))
        class_num = int(input("반: "))
        num = int(input("번호: "))
        new_std_id = (grade, class_num, num)

        if new_std_id in db:
            print(f"오류: {grade}학년 {class_num}반 {num}번 정보는 이미 존재합니다.")

        else:
            db[new_std_id] = {
                "이름": name,
                "학년": grade,
                "반": class_num,
                "번호": num,
                "성적": {"국어": 0, "영어": 0, "수학": 0},    # 중첩 딕셔너리
                "평균": 0.0
                }

            # 인덱스 업데이트
            if name not in name_index:
                name_index[name] = []
            name_index[name].append(new_std_id)
            print(f"추가 성공: {name} 학생의 정보가 추가되었습니다.")

    except ValueError:
        print("학년, 반, 번호는 숫자로 입력해야 합니다.")


# 학생 검색 함수
def search_student(db, name_index):
    search_name = input("\n검색할 학생 이름: ")
    if search_name in name_index:
        print(f"[{search_name}] 검색 결과:")
        for std_id in name_index[search_name]:
            info = db[std_id]
            print(f"- {info['학년']}학년 {info['반']}반 {info['번호']}번")
            print(f" 성적: 국어({info['성적']['국어']}), 영어({info['성적']['영어']}), 수학({info['성적']['수학']})")
            print(f" 평균: {info['평균']:.2f}")

    else:
        print("해당 이름의 학생이 없습니다.")


# 성적 처리 함수
def input_scores(db):
    print("\n--- 성적 입력 시스템 ---")

    try:
        grade = int(input("학년: "))
        class_num = int(input("반: "))
        num = int(input("번호: "))
        std_id = (grade, class_num, num)

        if std_id in db:
            kor = int(input("국어: "))
            eng = int(input("영어: "))
            math = int(input("수학: "))

            db[std_id]["성적"] = {"국어": kor, "영어": eng, "수학": math}
            db[std_id]["평균"] = (kor + eng + math) / 3
            print(f"성적 입력 완료. (평균: {db[std_id]['평균']:.2f})")
        else:
            print("해당 학번의 학생이 없습니다.")
    except ValueError:
        print("점수는 숫자로 입력하세요.")





# 메인 실행부 ------

if __name__ == "__main__":
    # 1. DB 초기화
    std_DB, name_index = init_db()
    print(f"시스템 초기화 완료: {len(std_DB)}명의 학생 데이터가 생성되었습니다.")

    # 2. 메뉴 반복
    while True:
        menu = input("\n[1]추가 [2]검색 [3] 성적입력 [Q]종료: ").upper()
        if menu == "1":
            add_student(std_DB, name_index)
        elif menu == "2":
            search_student(std_DB, name_index)
        elif menu == "3":
            input_scores(std_DB)
        elif menu == "Q":
            print("프로그램을 종료합니다.")
            break
        else:
            print("잘못된 입력입니다.")
```
