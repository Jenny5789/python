남은 문자열을 순서를 유지하여 배열로 return 하는 solution 함수를 완성해 주세요.

# 💻 Q3. X 사이의 개수

 myString을 문자 "x"를 기준으로 나눴을 때 나눠진 문자열 각각의 길이를 순서대로 저장한 배열을 return 하는 solution 함수를 완성해 주세요.



```python
myString = "oxooxoxxox"
```

```python
myString = "xabcxdefxghi"
```

```python
# "x"를 기준으로 나눴을 때
a = myString.split("x")
print(a)
```

```python
print(a.count())
```
