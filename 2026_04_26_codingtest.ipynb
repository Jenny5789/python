# 💻 Q3. 배열의 유사도
두 배열이 얼마나 유사한지 확인해보려고 합니다. 문자열 배열 s1과 s2가 주어질 때 같은 원소의 개수를 return하도록 solution 함수를 완성해주세요.

```python
s1 = ["a", "b", "c"]
s2 = ["com", "b", "d", "p", "c"]
```

```python
s1 = ["n", "omg"]
s2 = ["m", "dot"]
```

```python
"a"in s2
```

```python
"b"in s2
```

```python
"c"in s2
```

```python
for i in s1:
    print(i)
```

```python
# for문과 count()메소드 사용
count = 0
for i in s1:
    if i in s2:
        count = count + 1
print(count)
```

# 💻 Q4. n의 배수 고르기
정수 n과 정수 배열 numlist가 매개변수로 주어질 때, numlist에서 n의 배수가 아닌 수들을 제거한 배열을 return하도록 solution 함수를 완성해주세요.

```python
n = 3
numlist = [4, 5, 6, 7, 8, 9, 10, 11, 12]
```

```python
n = 5
numlist = [1, 9, 3, 10, 13, 5]
```

```python
n = 12
numlist = [2, 100, 120, 600, 12, 12]
```

```python
for i in numlist:
    print(i)
```

```python
if i % n == 0:
    print(i)
```

```python
result = []
for i in numlist:
    if i % n == 0:
        result.append(i)

print(result)
```
